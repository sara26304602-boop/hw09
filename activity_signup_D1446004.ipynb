# HW09 兩場活動報名整理器

本筆記本包含三組測試，分別為「測試一：原始資料」、「測試二：新增一位電影參加者」以及「測試三：兩場去重後完全相同」。

使用資料結構說明：
- `tuple` (`activity_names`): 儲存固定的活動名稱。
- `list` (`board_list`, `movie_list`): 儲存整理後（轉小寫、除空白）但保留重複項目的原始報名名單。
- `set` (`board_set`, `movie_set`): 用於去除重複報名，並進行集合運算（交集、聯集、差集）。
- `dict` (`unit_fees`, `summary`): `unit_fees` 用於查詢單價，`summary` 用於儲存各項人數與應收費用的計算結果。

## 測試一：原始資料

```python
# 1. 原始資料定義
activity_names = ("桌遊體驗", "電影欣賞")
board_raw = " Amy, Ben, amy, Cora "
movie_raw = "  ben, Dora, ERIC, ben "
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# 2. 讀取活動名稱
activity_board = activity_names[0]
activity_movie = activity_names[1]

# 3. 整理文字名單（移除空白、轉小寫、拆分串列）
# 處理桌遊名單
board_cleaned = board_raw.replace(" ", "").lower()
board_list = board_cleaned.split(",")

# 處理電影名單
movie_cleaned = movie_raw.replace(" ", "").lower()
movie_list = movie_cleaned.split(",")

# 4. 使用集合去重與運算
board_set = set(board_list)
movie_set = set(movie_list)

# 集合運算
common_participants = board_set & movie_set          # 共同參加者 (交集)
all_participants = board_set | movie_set             # 全部參加者 (聯集)
only_board = board_set - movie_set                   # 只參加桌遊者 (差集)
only_movie = movie_set - board_set                   # 只參加電影者 (差集)

# 排序名單以供顯示
board_sorted = sorted(list(board_set))
movie_sorted = sorted(list(movie_set))
common_sorted = sorted(list(common_participants))
all_sorted = sorted(list(all_participants))
only_board_sorted = sorted(list(only_board))
only_movie_sorted = sorted(list(only_movie))

# 5. 取得通知名單前兩位 (切片)
preview_list = all_sorted[:2]

# 6. 建立 summary 字典並計算人數與應收費用
summary = {}
summary["board_count"] = len(board_set)
summary["movie_count"] = len(movie_set)
summary["common_count"] = len(common_participants)
summary["total_unique_count"] = len(all_participants)
summary["total_registrations"] = len(board_set) + len(movie_set)

# 計算費用
summary["board_fee"] = summary["board_count"] * unit_fees[activity_board]
summary["movie_fee"] = summary["movie_count"] * unit_fees[activity_movie]
summary["total_fee"] = summary["board_fee"] + summary["movie_fee"]

# 7. 輸出完整摘要
print("HW09 活動報名摘要 (測試一)")
print(f"活動：{activity_board}，每人 {unit_fees[activity_board]:.2f} 元")
print(f"活動：{activity_movie}，每人 {unit_fees[activity_movie]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_sorted}")
print(f"電影去重名單：{movie_sorted}")
print(f"桌遊人數：{summary['board_count']} 人")
print(f"電影人數：{summary['movie_count']} 人")
print(f"共同參加者：{common_sorted}")
print(f"共同人數：{summary['common_count']} 人")
print(f"只參加桌遊者：{only_board_sorted}")
print(f"只參加電影者：{only_movie_sorted}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['total_unique_count']} 人")
print(f"報名人次：{summary['total_registrations']} 人次")
print(f"通知名單前兩位：{preview_list}")
print(f"桌遊應收：{summary['board_fee']:.2f} 元")
print(f"電影應收：{summary['movie_fee']:.2f} 元")
print(f"總應收：{summary['total_fee']:.2f} 元")
```
